# Personalized meal suitability — data readiness and label audit

This notebook does not generate data. It checks the real labelled dataset supplied by the researcher and stops if it is empty or missing provenance.

```python
from pathlib import Path
import pandas as pd
DATA = Path('../data/labelled/meal_suitability.csv')
REQUIRED = ['sodium_mg','potassium_mg','energy_kcal','sat_fat_g','na_k_ratio','age_years','bmi','dietary_requirements_count','suitability_label','source_dataset','source_record_id','participant_id','label_source']
print('Dataset path:', DATA.resolve())
df = pd.read_csv(DATA)
print('Rows:', len(df))
print('Missing required columns:', sorted(set(REQUIRED)-set(df.columns)))
if df.empty:
    print('STATUS: NOT READY — no labelled records. No model should be trained.')
else:
    print(df[REQUIRED].isna().sum())
    print('Label counts:')
    display(df['suitability_label'].value_counts(dropna=False))
    print('Label provenance:')
    display(df['label_source'].value_counts(dropna=False))
```

## Research gate

Before training, confirm label provenance, permitted use/consent, class definitions, participant grouping, duplicate records, missingness, and whether each label comes from expert review or a validated external dataset. Threshold-generated labels must not be described as clinical ground truth.
